## 🧰 ขั้นตอนที่ 2: เตรียมเครื่องมือและตรวจเครื่อง
ช่องนี้จะ (1) ติดตั้งส่วนเสริมที่ขาด (2) ตั้งค่าให้กราฟแสดง **ภาษาไทยได้ถูกต้อง** (3) ตรวจว่ามี GPU หรือไม่

**ผลที่ควรเห็น:** ข้อความ `✅ พร้อมใช้ GPU` พร้อมชื่อการ์ดจอ (เช่น Tesla T4)
ถ้าขึ้น `⚠️ ใช้ CPU` แปลว่ายังไม่ได้เปิด GPU → กลับไปทำข้อ 1 ในส่วน "วิธีใช้" (จะช้ากว่าประมาณ 10–20 เท่า)

````python
#@title 🧰 เตรียมเครื่องมือ + ตั้งค่าฟอนต์ไทย + ตรวจ GPU { display-mode: "form" }
import os, sys, re, json, copy, time, glob, random, shutil, zipfile, subprocess, urllib.request, warnings
from pathlib import Path
from datetime import datetime
warnings.filterwarnings("ignore")

IN_COLAB = "google.colab" in sys.modules

# ---------- ฟังก์ชันพิมพ์ข้อความให้อ่านง่าย ----------
def ok(m):   print(f"✅ {m}")
def info(m): print(f"ℹ️  {m}")
def warn(m): print(f"⚠️  {m}")

class StopHere(Exception):
    # หยุดการทำงานพร้อมข้อความสั้นๆ ที่อ่านเข้าใจง่าย (ไม่แสดงรายละเอียดโค้ดที่น่าสับสน)
    def _render_traceback_(self):
        return ["⛔ " + str(self)]

# ---------- ติดตั้งส่วนเสริมที่ขาด ----------
def _have(mod):
    try:
        __import__(mod); return True
    except Exception:
        return False

_need = [pkg for mod, pkg in [("onnx", "onnx"), ("onnxruntime", "onnxruntime"), ("onnxscript", "onnxscript"),
                              ("sklearn", "scikit-learn"), ("tqdm", "tqdm"), ("pandas", "pandas")] if not _have(mod)]
if _need:
    info(f"กำลังติดตั้งส่วนเสริม: {', '.join(_need)} (ครั้งแรกอาจใช้เวลา 1-2 นาที)")
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", *_need], check=False)

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms
from PIL import Image, ImageOps, ImageDraw
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from tqdm.auto import tqdm
from IPython.display import display, clear_output
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, precision_recall_fscore_support, accuracy_score

# ---------- ฟอนต์ภาษาไทยสำหรับกราฟ ----------
THAI_FONTS = ["Sarabun", "Noto Sans Thai", "Loma", "Garuda", "Norasi", "Waree", "Kinnari", "Umpush", "Tlwg Typist"]
def _find_thai_font():
    names = {f.name for f in fm.fontManager.ttflist}
    for c in THAI_FONTS:
        if c in names:
            return c
    return None

_thai = _find_thai_font()
if _thai is None and IN_COLAB:
    info("กำลังติดตั้งฟอนต์ภาษาไทยสำหรับกราฟ ...")
    subprocess.run("apt-get -qq install -y fonts-thai-tlwg > /dev/null 2>&1", shell=True)
    for f in glob.glob("/usr/share/fonts/**/*.[ot]tf", recursive=True):
        if "tlwg" in f.lower():
            try: fm.fontManager.addfont(f)
            except Exception: pass
    _thai = _find_thai_font()
if _thai:
    matplotlib.rcParams["font.family"] = _thai
    THAI_OK = True
else:
    THAI_OK = False
    warn("ไม่พบฟอนต์ไทย กราฟจะใช้ชื่อภาษาอังกฤษแทน (ไม่กระทบการฝึกโมเดล)")
matplotlib.rcParams["axes.unicode_minus"] = False
matplotlib.rcParams["figure.dpi"] = 110

# ---------- ตั้งค่าเลขสุ่มให้ผลซ้ำได้ ----------
def set_seed(s):
    random.seed(s); np.random.seed(s); torch.manual_seed(s)
    if torch.cuda.is_available(): torch.cuda.manual_seed_all(s)
set_seed(SEED)

# ---------- ตรวจอุปกรณ์ ----------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
USE_AMP = device.type == "cuda"   # ใช้การคำนวณแบบประหยัดหน่วยความจำ/เร็วขึ้นบน GPU
print("-" * 60)
print(f"Python {sys.version.split()[0]} | PyTorch {torch.__version__} | torchvision {__import__('torchvision').__version__}")
if device.type == "cuda":
    p = torch.cuda.get_device_properties(0)
    ok(f"พร้อมใช้ GPU: {p.name} (หน่วยความจำ {p.total_memory/1024**3:.1f} GB)")
else:
    warn("ตอนนี้ใช้ CPU ซึ่งช้ามาก  → เมนู รันไทม์ > เปลี่ยนประเภทรันไทม์ > T4 GPU แล้วรันใหม่ตั้งแต่ต้น")
print("-" * 60)
````

## 📥 ขั้นตอนที่ 4: โหลดรูปภาพ และตรวจสอบคุณภาพข้อมูล
ช่องนี้จะ **หาโฟลเดอร์รูปเอง** ไม่ต้องระบุตำแหน่ง ตรวจไฟล์รูปเสียและรายงานจำนวนรูปแต่ละโรค

**โครงสร้างที่ต้องมี (กรณีใช้รูปของตัวเอง):** หนึ่งโฟลเดอร์ต่อหนึ่งโรค ชื่อโฟลเดอร์ = ชื่อโรค
```
dataset/
 ├─ Algal Leaf Spot/   รูป1.jpg รูป2.jpg ...
 ├─ Leaf Blight/       ...
 ├─ Leaf Spot/         ...
 └─ Healthy/           ...
```
**เกณฑ์ที่ดี:** อย่างน้อย ~100 รูปต่อโรค และจำนวนรูปแต่ละโรคไม่ต่างกันมากเกินไป ระบบจะเตือนให้อัตโนมัติ

````python
#@title 📥 โหลด + ตรวจสอบรูปภาพ { display-mode: "form" }
IMG_EXT = {".jpg", ".jpeg", ".png", ".bmp", ".webp", ".tif", ".tiff"}

def has_images(folder):
    try:
        return any(p.is_file() and p.suffix.lower() in IMG_EXT for p in Path(folder).iterdir())
    except Exception:
        return False

def find_dataset_root(top):
    # หาโฟลเดอร์ที่มีโฟลเดอร์ย่อย (=ชื่อโรค) ซึ่งข้างในมีรูปมากที่สุด
    best_n, best_dir = 0, None
    for dp, dns, _ in os.walk(top):
        dns[:] = [d for d in dns if not d.startswith(".") and d != "__MACOSX"]
        n = sum(1 for d in dns if has_images(Path(dp) / d))
        if n > best_n:
            best_n, best_dir = n, Path(dp)
    return best_dir if best_n >= 2 else None

def safe_extract(zip_path, out_dir):
    out_dir = Path(out_dir).resolve(); out_dir.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(zip_path) as z:
        for m in z.infolist():
            target = (out_dir / m.filename).resolve()
            if not str(target).startswith(str(out_dir)):
                raise StopHere("ไฟล์ ZIP มีเส้นทางไม่ปลอดภัย จึงไม่แตกไฟล์")
        z.extractall(out_dir)

def make_demo_dataset(root, per_class=40, size=256):
    # สร้างรูปจำลอง (ไม่ใช่ใบทุเรียนจริง) เพื่อทดสอบว่าระบบทำงานครบทุกขั้นตอนเท่านั้น
    rng = np.random.default_rng(SEED)
    for cls in ["Algal Leaf Spot", "Leaf Blight", "Leaf Spot", "Healthy"]:
        d = Path(root) / cls; d.mkdir(parents=True, exist_ok=True)
        for i in range(per_class):
            base = np.clip(rng.normal([50, 130, 50], 18, (size, size, 3)), 0, 255).astype(np.uint8)
            im = Image.fromarray(base); dr = ImageDraw.Draw(im)
            for _ in range(int(rng.integers(8, 16))):
                x, y = (int(v) for v in rng.integers(20, size - 20, 2)); r = int(rng.integers(4, 10))
                if cls == "Algal Leaf Spot":  dr.ellipse([x-r, y-r, x+r, y+r], fill=(200, 140, 40))
                elif cls == "Leaf Blight":    dr.rectangle([x-2*r, y-r, x+2*r, y+r], fill=(120, 70, 30))
                elif cls == "Leaf Spot":      dr.ellipse([x-r, y-r, x+r, y+r], outline=(60, 30, 20), width=3)
            im.save(d / f"demo_{i:03d}.jpg", quality=90)

# ---------- 1) เลือกแหล่งข้อมูล ----------
WORK_DATA = Path("work_data").resolve()
data_root = None

if "Local" in DATA_SOURCE or "ในเครื่อง" in DATA_SOURCE or (Path("dataset").exists() and has_images(Path("dataset") / "Healthy")):
    for cand in [Path("dataset"), Path("../dataset"), Path("Durian_Leaf_Disease_AI/dataset")]:
        if cand.exists():
            found = find_dataset_root(cand)
            if found:
                data_root = found
                break
    if data_root is None and not ("จำลอง" in DATA_SOURCE):
        info("ไม่พบโฟลเดอร์ dataset ในเครื่อง จะลองตรวจสอบตามตัวเลือกอื่น...")

if "จำลอง" in DATA_SOURCE:
    warn("โหมดข้อมูลจำลอง: ใช้ทดสอบระบบเท่านั้น โมเดลที่ได้ตรวจใบทุเรียนจริงไม่ได้")
    make_demo_dataset(WORK_DATA / "demo")
    data_root = WORK_DATA / "demo"

elif "GitHub" in DATA_SOURCE or "Roboflow" in DATA_SOURCE or "ดาวน์โหลด" in DATA_SOURCE:
    zip_path = Path("durian_dataset.zip")
    extract_dir = WORK_DATA / "github"
    try:
        if not zip_path.exists():
            info("กำลังดาวน์โหลดชุดข้อมูลจาก GitHub (ขึ้นกับความเร็วเน็ต อาจใช้เวลาหลายนาที) ...")
            req = urllib.request.Request(DATASET_URL, headers={"User-Agent": "Mozilla/5.0"})
            with urllib.request.urlopen(req, timeout=120) as r, open(zip_path, "wb") as f:
                shutil.copyfileobj(r, f)
        info(f"ดาวน์โหลดแล้ว ({zip_path.stat().st_size/1024**2:.1f} MB) กำลังแตกไฟล์ ...")
        safe_extract(zip_path, extract_dir)
        data_root = find_dataset_root(extract_dir)
    except StopHere:
        raise
    except Exception as e:
        warn(f"ดาวน์โหลดไม่สำเร็จ: {e}")
    if data_root is None:
        raise StopHere("ไม่พบรูปภาพที่จัดเป็นโฟลเดอร์ตามโรคจากลิงก์นี้\n"
                       "   วิธีแก้: ข้อ 1) เปลี่ยน 'แหล่งข้อมูล' เป็น 'โฟลเดอร์ในไดรฟ์ของฉัน' หรือ 'อัปโหลดไฟล์ ZIP'\n"
                       "            ข้อ 2) หรือเปลี่ยนลิงก์ DATASET_URL เป็นของชุดข้อมูลที่ถูกต้อง แล้วรันช่องตั้งค่าใหม่")

elif "Google Drive" in DATA_SOURCE:
    folder = BASE_DIR / "dataset"
    folder.mkdir(parents=True, exist_ok=True)
    data_root = find_dataset_root(folder)
    if data_root is None:
        raise StopHere(f"ยังไม่พบรูปภาพในโฟลเดอร์: {folder}\n"
                       "   วิธีแก้: อัปโหลดโฟลเดอร์รูปแยกตามโรคเข้าไปที่ตำแหน่งนี้ใน Google Drive แล้วกดรันช่องนี้ใหม่")

else:  # อัปโหลด ZIP
    if not IN_COLAB:
        raise StopHere("ตัวเลือกอัปโหลด ZIP ใช้ได้บน Google Colab เท่านั้น")
    from google.colab import files
    print("👇 กดปุ่ม 'Choose Files' แล้วเลือกไฟล์ .zip ที่มีโฟลเดอร์รูปแยกตามโรค")
    up = files.upload()
    zips = [n for n in up if n.lower().endswith(".zip")]
    if not zips:
        raise StopHere("ไม่พบไฟล์ .zip ที่อัปโหลด กรุณารันช่องนี้ใหม่แล้วเลือกไฟล์ ZIP")
    safe_extract(zips[0], WORK_DATA / "upload")
    data_root = find_dataset_root(WORK_DATA / "upload")
    if data_root is None:
        raise StopHere("ใน ZIP ไม่พบโฟลเดอร์รูปที่แยกตามโรค (ต้องมีอย่างน้อย 2 โฟลเดอร์ที่มีรูปอยู่ข้างใน)")

ok(f"พบชุดข้อมูลที่: {data_root}")

# ---------- 2) รวบรวมรายการรูป + ตรวจไฟล์เสีย ----------
class_names = sorted(d.name for d in Path(data_root).iterdir()
                     if d.is_dir() and not d.name.startswith(".") and d.name != "__MACOSX" and has_images(d))
NUM_CLASSES = len(class_names)
class_to_idx = {c: i for i, c in enumerate(class_names)}

all_paths, all_labels, bad_files = [], [], []
for c in class_names:
    files_c = sorted(p for p in (Path(data_root) / c).iterdir() if p.is_file() and p.suffix.lower() in IMG_EXT)
    for p in tqdm(files_c, desc=f"ตรวจรูป {c}", leave=False):
        try:
            with Image.open(p) as im:
                im.verify()
            all_paths.append(str(p)); all_labels.append(class_to_idx[c])
        except Exception:
            bad_files.append(str(p))
all_labels = np.array(all_labels); all_paths = np.array(all_paths)

if bad_files:
    warn(f"พบไฟล์รูปเสีย {len(bad_files)} ไฟล์ ระบบข้ามให้อัตโนมัติ (เช่น {Path(bad_files[0]).name})")

counts = np.bincount(all_labels, minlength=NUM_CLASSES)
if counts.min() < 10:
    raise StopHere(f"โรค '{class_names[int(counts.argmin())]}' มีรูปเพียง {counts.min()} รูป (ต้องมีอย่างน้อย 10 รูป) กรุณาเพิ่มรูปแล้วรันใหม่")

# ---------- 3) ตารางสรุป + กราฟ ----------
df_cls = pd.DataFrame({
    "ลำดับ": range(1, NUM_CLASSES + 1),
    "ชื่อโรค (โฟลเดอร์)": class_names,
    "ชื่อไทย": [info_of(c)["th"] for c in class_names],
    "จำนวนรูป": counts,
    "สัดส่วน (%)": np.round(counts / counts.sum() * 100, 1),
})
print(f"\n📊 พบรูปที่ใช้ได้ทั้งหมด {len(all_paths):,} รูป ใน {NUM_CLASSES} กลุ่ม")
display(df_cls.style.hide(axis="index"))

fig, ax = plt.subplots(figsize=(8, 3.6))
bars = ax.barh([th_name(c) for c in class_names][::-1], counts[::-1], color="#2e8b57")
for b, v in zip(bars, counts[::-1]):
    ax.text(v, b.get_y() + b.get_height()/2, f" {v}", va="center")
ax.set_xlabel("จำนวนรูป" if THAI_OK else "images"); ax.set_title("จำนวนรูปภาพของแต่ละโรค" if THAI_OK else "Images per class")
ax.spines[["top", "right"]].set_visible(False); plt.tight_layout(); save_fig("01_class_distribution"); plt.show()

# ---------- 4) คำเตือนอัตโนมัติ ----------
if "Local" in DATA_SOURCE or "ในเครื่อง" in DATA_SOURCE or (Path("dataset").exists() and has_images(Path("dataset") / "Healthy")):
    for cand in [Path("dataset"), Path("../dataset"), Path("Durian_Leaf_Disease_AI/dataset")]:
        if cand.exists():
            found = find_dataset_root(cand)
            if found:
                data_root = found
                break
    if data_root is None and not ("จำลอง" in DATA_SOURCE):
        info("ไม่พบโฟลเดอร์ dataset ในเครื่อง จะลองตรวจสอบตามตัวเลือกอื่น...")

if "จำลอง" in DATA_SOURCE: warn("นี่คือข้อมูลจำลอง ห้ามนำโมเดลไปใช้ตรวจใบจริง")
if len(all_paths) < 300: warn(f"รูปทั้งหมดมีเพียง {len(all_paths)} รูป ซึ่งค่อนข้างน้อย ผลที่ได้อาจไม่เสถียร แนะนำให้เพิ่มรูปถ่ายจริงจากสวน")
ratio = counts.max() / counts.min()
if ratio > 3: warn(f"จำนวนรูปไม่สมดุล (มากสุด/น้อยสุด = {ratio:.1f} เท่า) ระบบจะถ่วงน้ำหนักการเรียนให้อัตโนมัติ")
if NUM_CLASSES != 4: info(f"พบ {NUM_CLASSES} กลุ่ม (ค่าตั้งต้นของโครงการคือ 4 กลุ่ม) ระบบรองรับจำนวนกลุ่มใดก็ได้")
ok("ตรวจสอบข้อมูลเสร็จสิ้น")
````

## 🖼️ ขั้นตอนที่ 5: ดูรูปตัวอย่างจริงด้วยตาตัวเอง
**ทำไมสำคัญ?** ถ้ารูปผิดโฟลเดอร์หรือรูปไม่ชัด ปัญญาประดิษฐ์จะเรียนผิดตามไปด้วย
**สิ่งที่ต้องทำ:** ดูว่าแต่ละแถว (หนึ่งโรค) รูปมีลักษณะแผลตรงกับชื่อโรคหรือไม่ ถ้าเห็นรูปผิดกลุ่ม ให้ย้ายไปโฟลเดอร์ที่ถูกต้องแล้วรันตั้งแต่ขั้นตอนที่ 4 ใหม่

````python
#@title 🖼️ แสดงรูปตัวอย่างของแต่ละโรค { display-mode: "form" }
def load_rgb(path, max_side=None):
    # เปิดรูป แก้การหมุนจากมือถือ (EXIF) และแปลงเป็นสี RGB; ลดขนาดตอนอ่านเพื่อให้เร็วขึ้น
    im = Image.open(path)
    if max_side and im.format == "JPEG":
        im.draft("RGB", (max_side, max_side))
    im = ImageOps.exif_transpose(im)
    return im.convert("RGB")

PER_ROW = 5
rng = np.random.default_rng(SEED)
fig, axes = plt.subplots(NUM_CLASSES, PER_ROW, figsize=(2.6 * PER_ROW, 2.6 * NUM_CLASSES), squeeze=False)
for r, c in enumerate(class_names):
    idxs = np.where(all_labels == r)[0]
    pick = rng.choice(idxs, size=min(PER_ROW, len(idxs)), replace=False)
    for k in range(PER_ROW):
        ax = axes[r][k]; ax.axis("off")
        if k < len(pick):
            ax.imshow(load_rgb(all_paths[pick[k]], max_side=512))
            if k == 0: ax.set_title(th_name(c), loc="left", fontsize=11, color="darkgreen", weight="bold")
plt.suptitle("ตัวอย่างรูปจริงในชุดข้อมูล (แถวละ 1 โรค)" if THAI_OK else "Sample images (one row per class)", y=1.0, fontsize=13)
plt.tight_layout(); save_fig("02_sample_images"); plt.show()
ok("แสดงรูปตัวอย่างแล้ว — ตรวจว่ารูปในแต่ละแถวตรงกับชื่อโรคหรือไม่")
````

## ✂️ ขั้นตอนที่ 6: แบ่งข้อมูล และสร้าง "ตัวแต่งรูป" (Data Augmentation)
**แบ่งข้อมูล 3 ชุด (แบบสุ่มแต่รักษาสัดส่วนโรคให้เท่ากันทุกชุด):**
- 🟢 **ชุดฝึก 70%** — ให้ปัญญาประดิษฐ์เรียน
- 🟡 **ชุดสอบย่อย 15%** — ใช้เช็กระหว่างเรียน และเลือกรอบที่ดีที่สุด
- 🔴 **ชุดสอบจริง 15%** — **เก็บไว้ ไม่ให้เห็นระหว่างเรียนเลย** ใช้วัดผลตอนจบ เพื่อให้ตัวเลขน่าเชื่อถือ

**ตัวแต่งรูป:** ตอนฝึกจะสุ่มครอบ พลิก หมุน และปรับแสง/สีเล็กน้อย เพื่อให้ปัญญาประดิษฐ์ทนต่อมุมกล้องและแสงต่างๆ ในสวนจริง (ชุดสอบจะไม่แต่งรูป)

````python
#@title ✂️ แบ่งข้อมูล + สร้างตัวแต่งรูป + ตัวป้อนข้อมูล { display-mode: "form" }
MEAN, STD = [0.485, 0.456, 0.406], [0.229, 0.224, 0.225]   # ค่ามาตรฐาน ImageNet ที่โมเดลต้นแบบคุ้นเคย

train_tf = transforms.Compose([
    transforms.RandomResizedCrop(IMAGE_SIZE, scale=(0.6, 1.0)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomVerticalFlip(p=0.2),
    transforms.RandomRotation(25),
    transforms.ColorJitter(brightness=0.25, contrast=0.25, saturation=0.2, hue=0.02),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])
eval_tf = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(MEAN, STD),
])

class LeafDataset(Dataset):
    def __init__(self, paths, labels, transform):
        self.paths, self.labels, self.transform = list(paths), list(labels), transform
    def __len__(self):
        return len(self.paths)
    def __getitem__(self, i):
        img = load_rgb(self.paths[i], max_side=IMAGE_SIZE * 2)
        return self.transform(img), int(self.labels[i])

# แบ่ง 70 / 15 / 15 แบบรักษาสัดส่วนโรค (stratified)
idx_all = np.arange(len(all_paths))
tr_idx, tmp_idx = train_test_split(idx_all, test_size=0.30, stratify=all_labels, random_state=SEED)
va_idx, te_idx  = train_test_split(tmp_idx, test_size=0.50, stratify=all_labels[tmp_idx], random_state=SEED)

train_ds = LeafDataset(all_paths[tr_idx], all_labels[tr_idx], train_tf)
val_ds   = LeafDataset(all_paths[va_idx], all_labels[va_idx], eval_tf)
test_ds  = LeafDataset(all_paths[te_idx], all_labels[te_idx], eval_tf)

NUM_WORKERS = 2 if IN_COLAB else 0
g = torch.Generator(); g.manual_seed(SEED)
loader_kw = dict(batch_size=BATCH_SIZE, num_workers=NUM_WORKERS, pin_memory=(device.type == "cuda"))
train_loader = DataLoader(train_ds, shuffle=True, drop_last=len(train_ds) > BATCH_SIZE, generator=g, **loader_kw)
val_loader   = DataLoader(val_ds, shuffle=False, **loader_kw)
test_loader  = DataLoader(test_ds, shuffle=False, **loader_kw)

# ถ่วงน้ำหนักโรคที่มีรูปน้อย เพื่อไม่ให้ปัญญาประดิษฐ์ลำเอียงไปทางโรคที่มีรูปเยอะ
tr_counts = np.bincount(all_labels[tr_idx], minlength=NUM_CLASSES)
class_weights = torch.tensor(tr_counts.sum() / (NUM_CLASSES * np.maximum(tr_counts, 1)), dtype=torch.float32)

split_df = pd.DataFrame({
    "ชุดข้อมูล": ["🟢 ชุดฝึก", "🟡 ชุดสอบย่อย", "🔴 ชุดสอบจริง", "รวม"],
    "จำนวนรูป": [len(train_ds), len(val_ds), len(test_ds), len(all_paths)],
})
display(split_df.style.hide(axis="index"))
ok("แบ่งข้อมูลเรียบร้อย (รูปแต่ละใบอยู่ได้เพียงชุดเดียว จึงไม่มีการ 'แอบดูข้อสอบ')")

# ---------- แสดงตัวอย่างการแต่งรูป ----------
k = int(np.random.default_rng(SEED).integers(len(train_ds)))
src_img = load_rgb(train_ds.paths[k], max_side=IMAGE_SIZE * 2)
def _denorm(t):
    a = t.numpy().transpose(1, 2, 0) * np.array(STD) + np.array(MEAN)
    return np.clip(a, 0, 1)
fig, axes = plt.subplots(2, 4, figsize=(11, 5.6))
axes[0][0].imshow(src_img.resize((IMAGE_SIZE, IMAGE_SIZE))); axes[0][0].set_title("รูปต้นฉบับ" if THAI_OK else "original", color="darkgreen")
for j, ax in enumerate(axes.flat[1:], 1):
    ax.imshow(_denorm(train_tf(src_img))); ax.set_title(f"แต่งแบบที่ {j}" if THAI_OK else f"augmented {j}", fontsize=9)
for ax in axes.flat: ax.axis("off")
plt.suptitle(f"รูปเดียวกัน แต่งต่างกัน 7 แบบ (รูปนี้คือ: {th_name(class_names[train_ds.labels[k]])})" if THAI_OK else "One image, 7 augmentations", fontsize=12)
plt.tight_layout(); save_fig("03_augmentation"); plt.show()
````

## 🧠 ขั้นตอนที่ 7: สร้างโมเดล (Transfer Learning)
**หลักการ:** ใช้โมเดลที่เคยเรียนรู้จากรูปนับล้าน (ImageNet) มาแล้ว เปลี่ยน "ชั้นตอบคำถามสุดท้าย" ให้ตอบเป็นโรคใบทุเรียนของเรา — ไม่ต้องเริ่มสอนจากศูนย์ จึงใช้รูปไม่มากและฝึกเร็ว

**กลยุทธ์ฝึก 2 ช่วง (วิธีมาตรฐานที่ให้ผลดี):**
1. **ช่วงฝึกส่วนหัว** — ล็อกความรู้เดิมไว้ ฝึกเฉพาะชั้นตอบคำถามใหม่ให้พอใช้ก่อน
2. **ช่วงปรับจูน** — ปลดล็อกทั้งโมเดล ปรับละเอียดด้วยอัตราเรียนที่ค่อยๆ ลดลง

ถ้ากังวลเรื่อง `weights` ดาวน์โหลดไม่ได้ ระบบจะแจ้งเตือนให้ทราบ (ปกติบน Colab ดาวน์โหลดได้)

````python
#@title 🧠 สร้างโมเดลจากโมเดลต้นแบบ { display-mode: "form" }
def build_model(name, num_classes, pretrained=True):
    # คืนค่า (โมเดล, ใช้น้ำหนักต้นแบบได้หรือไม่)
    spec = {
        "mobilenet_v3_small": (models.mobilenet_v3_small, models.MobileNet_V3_Small_Weights, 3),
        "mobilenet_v3_large": (models.mobilenet_v3_large, models.MobileNet_V3_Large_Weights, 3),
        "efficientnet_b0":    (models.efficientnet_b0,    models.EfficientNet_B0_Weights,    1),
    }
    if name not in spec:
        raise StopHere(f"ไม่รู้จักโมเดล '{name}' (เลือกได้: {', '.join(spec)})")
    fn, W, head_idx = spec[name]
    used_pretrained = False
    m = None
    if pretrained:
        try:
            m = fn(weights=W.DEFAULT); used_pretrained = True
        except Exception as e:
            warn(f"โหลดน้ำหนักต้นแบบไม่ได้ ({type(e).__name__}) → จะเริ่มเรียนจากศูนย์ ซึ่งต้องใช้รูปและเวลามากกว่า")
    if m is None:
        m = fn(weights=None)
    in_f = m.classifier[head_idx].in_features
    m.classifier[head_idx] = nn.Linear(in_f, num_classes)
    # MobileNetV3 ตั้งค่า BatchNorm ให้ปรับตัวช้ามาก (momentum 0.01) ทำให้ชุดข้อมูลเล็กๆ ปรับค่าไม่ทัน
    # จึงใช้ค่ามาตรฐาน 0.1 เพื่อให้ผลตอนสอบ/ใช้งานจริงตรงกับตอนฝึก
    for mod in m.modules():
        if isinstance(mod, nn.BatchNorm2d):
            mod.momentum = 0.1
    return m, used_pretrained

def set_backbone_trainable(m, flag):
    for p in m.features.parameters():
        p.requires_grad = flag

model, PRETRAINED_OK = build_model(MODEL_NAME, NUM_CLASSES)
model = model.to(device)

n_total = sum(p.numel() for p in model.parameters())
n_head  = sum(p.numel() for p in model.classifier.parameters())
ok(f"สร้างโมเดล {MODEL_NAME} สำเร็จ ({'ใช้ความรู้เดิมจาก ImageNet' if PRETRAINED_OK else 'เริ่มจากศูนย์'})")
print(f"   • พารามิเตอร์ทั้งหมด  : {n_total/1e6:.2f} ล้านตัว  (ไฟล์ประมาณ {n_total*4/1024**2:.1f} MB)")
print(f"   • ส่วนหัวที่ต้องเรียนใหม่: {n_head/1e6:.2f} ล้านตัว")
print(f"   • ตอบได้ {NUM_CLASSES} กลุ่ม: " + ", ".join(th_name(c) for c in class_names))
````

## 🏋️ ขั้นตอนที่ 8: ฝึกสอนปัญญาประดิษฐ์ (ดูผลสดๆ)
กดรันแล้ว **รอและดู** — ทุกครั้งที่จบ 1 รอบ จะมี:
- 📈 **กราฟ 2 รูปที่อัปเดตสดๆ** ซ้าย = ค่าความผิดพลาด (**ยิ่งต่ำยิ่งดี**) · ขวา = ความแม่นยำ (**ยิ่งสูงยิ่งดี**)
- 📝 **บรรทัดสรุปผลทุกรอบ** พร้อมเวลาที่เหลือโดยประมาณ
- ⭐ **ดาว** เมื่อได้ผลสอบย่อยดีที่สุดใหม่ ระบบจะบันทึกโมเดลรอบนั้นให้อัตโนมัติ

**วิธีอ่านกราฟ:** เส้นฟ้า (เรียน) กับเส้นส้ม (สอบย่อย) ควรเดินไปด้วยกัน ถ้าเส้นฟ้าดีขึ้นเรื่อยๆ แต่เส้นส้มแย่ลง แปลงว่า "ท่องจำ" (overfitting) — ระบบจะหยุดเองเมื่อไม่พัฒนาติดต่อกัน 8 รอบ
> ⏱️ บน T4 GPU ปกติรอบละ ~10–40 วินาที ห้ามปิดหน้าเว็บระหว่างฝึก

````python
#@title 🏋️ เริ่มฝึกสอน (รอจนขึ้น 🎉) { display-mode: "form" }
criterion_train = nn.CrossEntropyLoss(weight=class_weights.to(device), label_smoothing=0.1)
criterion_eval  = nn.CrossEntropyLoss()
scaler = torch.amp.GradScaler("cuda", enabled=USE_AMP)

def run_epoch(m, loader, optimizer=None, frozen=False, desc=""):
    training = optimizer is not None
    m.train(training)
    if training and frozen:
        m.features.eval()          # ช่วงล็อกความรู้เดิม: ไม่ให้ค่าสถิติภายในของโมเดลเปลี่ยน
    tot_loss, correct, n = 0.0, 0, 0
    with torch.set_grad_enabled(training):
        for x, y in tqdm(loader, desc=desc, leave=False):
            x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)
            if training: optimizer.zero_grad(set_to_none=True)
            with torch.autocast(device_type=device.type, dtype=torch.float16, enabled=USE_AMP):
                out = m(x)
                loss = (criterion_train if training else criterion_eval)(out, y)
            if training:
                scaler.scale(loss).backward(); scaler.step(optimizer); scaler.update()
            tot_loss += loss.item() * x.size(0)
            correct += (out.argmax(1) == y).sum().item()
            n += x.size(0)
    return tot_loss / max(n, 1), correct / max(n, 1) * 100

history = {"epoch": [], "train_loss": [], "val_loss": [], "train_acc": [], "val_acc": [], "phase": []}
log_lines, best = [], {"score": (-1.0, 0.0), "epoch": 0}
BEST_PATH = MODEL_DIR / "durian_disease_model_best.pth"
TOTAL_EPOCHS = EPOCHS_FROZEN + EPOCHS_FINETUNE

def draw_live():
    clear_output(wait=True)
    print(f"🏋️ กำลังฝึก {MODEL_NAME} | อุปกรณ์: {device} | ทั้งหมด {TOTAL_EPOCHS} รอบ (หยุดเองได้ถ้าไม่พัฒนา)")
    print("-" * 100)
    for ln in log_lines: print(ln)
    print("-" * 100)
    ep = history["epoch"]
    fig, ax = plt.subplots(1, 2, figsize=(12, 3.8))
    for a, (k1, k2, ttl) in zip(ax, [("train_loss", "val_loss", "ค่าความผิดพลาด (ยิ่งต่ำยิ่งดี)"), ("train_acc", "val_acc", "ความแม่นยำ % (ยิ่งสูงยิ่งดี)")]):
        a.plot(ep, history[k1], "o-", color="#1f77b4", label="เรียน (ชุดฝึก)")
        a.plot(ep, history[k2], "o-", color="#ff7f0e", label="สอบย่อย")
        if len(ep) > EPOCHS_FROZEN: a.axvline(EPOCHS_FROZEN + 0.5, ls=":", color="gray")
        if best["epoch"]: a.axvline(best["epoch"], ls="--", color="green", alpha=.5)
        a.set_title(ttl); a.set_xlabel("รอบที่"); a.grid(alpha=.3); a.legend()
    plt.tight_layout(); plt.show()

t_start = time.time(); no_improve = 0; stopped_early = False; epoch_no = 0
phases = [("ฝึกส่วนหัว", EPOCHS_FROZEN, True), ("ปรับจูนทั้งโมเดล", EPOCHS_FINETUNE, False)]

for phase_name, n_ep, frozen in phases:
    set_backbone_trainable(model, not frozen)
    params = [p for p in model.parameters() if p.requires_grad]
    optimizer = torch.optim.AdamW(params, lr=(HEAD_LR if frozen else LEARNING_RATE), weight_decay=1e-4)
    scheduler = None if frozen else torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=max(n_ep, 1))
    for _ in range(n_ep):
        epoch_no += 1; t0 = time.time()
        tr_loss, tr_acc = run_epoch(model, train_loader, optimizer, frozen, desc=f"รอบ {epoch_no}/{TOTAL_EPOCHS} เรียน")
        va_loss, va_acc = run_epoch(model, val_loader, None, desc=f"รอบ {epoch_no}/{TOTAL_EPOCHS} สอบย่อย")
        if scheduler: scheduler.step()

        for k, v in zip(history, [epoch_no, tr_loss, va_loss, tr_acc, va_acc, phase_name]): history[k].append(v)
        score = (va_acc, -va_loss)
        star = ""
        if score > best["score"]:
            best.update(score=score, epoch=epoch_no); no_improve = 0; star = "⭐ บันทึกโมเดลที่ดีที่สุด"
            torch.save({
                "epoch": epoch_no, "model_state_dict": model.state_dict(), "classes": class_names,
                "architecture": MODEL_NAME, "image_size": IMAGE_SIZE, "mean": MEAN, "std": STD,
                "val_acc": va_acc, "val_loss": va_loss,
            }, BEST_PATH)
        else:
            no_improve += 1
        sec = time.time() - t0
        left = (TOTAL_EPOCHS - epoch_no) * sec / 60
        log_lines.append(f"รอบ {epoch_no:02d}/{TOTAL_EPOCHS} [{phase_name}] เรียน {tr_acc:5.1f}% (ผิดพลาด {tr_loss:.3f}) | สอบย่อย {va_acc:5.1f}% (ผิดพลาด {va_loss:.3f}) | {sec:.0f} วินาที | เหลือ ~{left:.1f} นาที {star}")
        draw_live()
        ft_done = epoch_no - EPOCHS_FROZEN
        if not frozen and ft_done >= 8 and no_improve >= PATIENCE:   # ให้ฝึกอย่างน้อย 8 รอบปรับจูนก่อนจึงจะหยุดเองได้
            stopped_early = True; break
    if stopped_early: break

mins = (time.time() - t_start) / 60
if stopped_early: info(f"หยุดฝึกก่อนกำหนดที่รอบ {epoch_no} เพราะผลสอบไม่ดีขึ้นติดกัน {PATIENCE} รอบ (เป็นเรื่องปกติและดี ช่วยกันท่องจำ)")
ok(f"🎉 ฝึกเสร็จสมบูรณ์ ใช้เวลา {mins:.1f} นาที | ดีที่สุดที่รอบ {best['epoch']} (ความแม่นยำสอบย่อย {best['score'][0]:.1f}%)")
ok(f"บันทึกโมเดลที่ดีที่สุดไว้ที่: {BEST_PATH}")

# โหลดโมเดลรอบที่ดีที่สุดกลับมาใช้ต่อ
ckpt = torch.load(BEST_PATH, map_location=device)
model.load_state_dict(ckpt["model_state_dict"]); model.eval()
set_backbone_trainable(model, True)
````

## 📈 ขั้นตอนที่ 9: สรุปการเรียนรู้ และตรวจสุขภาพของการฝึก
ช่องนี้สรุปทุกรอบเป็นตาราง และ **วินิจฉัยให้อัตโนมัติ** ว่าการฝึกครั้งนี้ปกติหรือไม่ พร้อมคำแนะนำภาษาไทย

````python
#@title 📈 ตารางสรุปทุกรอบ + วินิจฉัยการฝึก { display-mode: "form" }
hist_df = pd.DataFrame({
    "รอบที่": history["epoch"], "ช่วง": history["phase"],
    "เรียน (%)": np.round(history["train_acc"], 1), "สอบย่อย (%)": np.round(history["val_acc"], 1),
    "ผิดพลาดตอนเรียน": np.round(history["train_loss"], 3), "ผิดพลาดตอนสอบ": np.round(history["val_loss"], 3),
})
hist_df.insert(2, "ดีที่สุด", ["⭐" if e == best["epoch"] else "" for e in history["epoch"]])
display(hist_df.style.hide(axis="index"))
hist_df.to_csv(REPORT_DIR / "training_history.csv", index=False, encoding="utf-8-sig")

bi = best["epoch"] - 1
tr_a, va_a = history["train_acc"][bi], history["val_acc"][bi]
gap = tr_a - va_a
print("\n🩺 ผลวินิจฉัยการฝึก")
if va_a < 70:
    warn(f"ความแม่นยำสอบย่อยเพียง {va_a:.1f}% — โมเดลยังเรียนไม่พอ\n   ลอง: เพิ่ม 'รอบปรับจูน', เพิ่มรูปให้มากขึ้น, หรือตรวจว่ารูปอยู่ถูกโฟลเดอร์")
elif gap > 15:
    warn(f"เรียนได้ {tr_a:.1f}% แต่สอบย่อยได้ {va_a:.1f}% (ห่าง {gap:.1f}%) — เริ่ม 'ท่องจำ' (overfitting)\n   ลอง: เพิ่มรูปจริงหลากหลายมุม/แสง, ลดจำนวนรอบ")
else:
    ok(f"การฝึกอยู่ในเกณฑ์ดี: เรียน {tr_a:.1f}% / สอบย่อย {va_a:.1f}% (ห่างกัน {gap:+.1f}%)")
if not PRETRAINED_OK:
    warn("ครั้งนี้ไม่ได้ใช้ความรู้เดิมจาก ImageNet ผลอาจต่ำกว่าปกติ → รันใหม่เมื่อเน็ตเสถียร")
````

## 🎯 ขั้นตอนที่ 10: สอบจริง! (ประเมินด้วยรูปที่ไม่เคยเห็น)
นี่คือ **ตัวเลขที่ควรรายงาน** เพราะใช้ชุดสอบจริงที่ปัญญาประดิษฐ์ไม่เคยเห็นเลย

**วิธีอ่านผล:**
- **ความแม่นยำรวม** — ตอบถูกกี่ % ของทั้งหมด
- **Precision (ความเที่ยงตรง)** — เมื่อบอกว่า "เป็นโรค X" ถูกจริงกี่ %
- **Recall (ความไว)** — จากใบที่เป็นโรค X จริงๆ จับได้กี่ % *(ในงานเกษตร ค่านี้สำคัญ เพราะพลาดโรค = ลุกลาม)*
- **F1** — ค่ากลางที่รวมสองตัวบน (ยิ่งใกล้ 100 ยิ่งดี)
- **ตารางความสับสน (Confusion Matrix)** — แถว = ความจริง, คอลัมน์ = สิ่งที่ปัญญาประดิษฐ์ตอบ เส้นทแยงมุมเข้มๆ = ตอบถูก

````python
#@title 🎯 สอบจริง + ตารางความสับสน + รูปที่ตอบผิด { display-mode: "form" }
@torch.no_grad()
def predict_loader(m, loader):
    m.eval(); P, Y = [], []
    for x, y in tqdm(loader, desc="กำลังสอบจริง", leave=False):
        P.append(F.softmax(m(x.to(device)).float(), 1).cpu().numpy()); Y.append(y.numpy())
    return np.concatenate(P), np.concatenate(Y)

probs_te, y_te = predict_loader(model, test_loader)
pred_te = probs_te.argmax(1)
test_acc = accuracy_score(y_te, pred_te) * 100
prec, rec, f1, sup = precision_recall_fscore_support(y_te, pred_te, labels=range(NUM_CLASSES), zero_division=0)

print(f"🎯 ความแม่นยำบนชุดสอบจริง: {test_acc:.2f}%  (จาก {len(y_te)} รูป ตอบถูก {int((pred_te == y_te).sum())} รูป)\n")
rep_df = pd.DataFrame({
    "โรค": [th_name(c) for c in class_names], "ชื่อเดิม": class_names,
    "Precision (%)": np.round(prec * 100, 1), "Recall (%)": np.round(rec * 100, 1),
    "F1 (%)": np.round(f1 * 100, 1), "จำนวนรูปสอบ": sup,
})
display(rep_df.style.hide(axis="index"))
rep_df.to_csv(REPORT_DIR / "test_report.csv", index=False, encoding="utf-8-sig")

# ---------- ตารางความสับสน ----------
cm = confusion_matrix(y_te, pred_te, labels=range(NUM_CLASSES))
cm_pct = cm / np.maximum(cm.sum(1, keepdims=True), 1) * 100
labels_plot = [th_name(c) for c in class_names]
fig, axes = plt.subplots(1, 2, figsize=(13, 5.2))
for ax, M, ttl, fmt in [(axes[0], cm, "จำนวนรูป", "{:.0f}"), (axes[1], cm_pct, "เปอร์เซ็นต์ของแต่ละโรคจริง", "{:.0f}%")]:
    im = ax.imshow(M, cmap="Greens")
    ax.set_xticks(range(NUM_CLASSES)); ax.set_yticks(range(NUM_CLASSES))
    ax.set_xticklabels(labels_plot, rotation=30, ha="right"); ax.set_yticklabels(labels_plot)
    ax.set_xlabel("ปัญญาประดิษฐ์ตอบ"); ax.set_ylabel("ความจริง"); ax.set_title(ttl)
    for i in range(NUM_CLASSES):
        for j in range(NUM_CLASSES):
            ax.text(j, i, fmt.format(M[i, j]), ha="center", va="center", color="white" if M[i, j] > M.max() * .55 else "black", fontsize=11)
plt.tight_layout(); save_fig("04_confusion_matrix"); plt.show()

# ---------- รูปที่ตอบผิด ----------
wrong = np.where(pred_te != y_te)[0]
if len(wrong):
    show = wrong[np.argsort(-probs_te[wrong].max(1))][:8]     # เรียงจาก "มั่นใจแต่ผิด" มากที่สุดก่อน
    cols = min(4, len(show)); rows = int(np.ceil(len(show) / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(3.6 * cols, 3.9 * rows), squeeze=False)
    for ax in axes.flat: ax.axis("off")
    for ax, k in zip(axes.flat, show):
        ax.imshow(load_rgb(test_ds.paths[k], max_side=512))
        ax.set_title(f"จริง: {th_name(class_names[y_te[k]])}\nตอบ: {th_name(class_names[pred_te[k]])} ({probs_te[k].max()*100:.0f}%)", fontsize=9, color="firebrick")
    plt.suptitle(f"รูปที่ตอบผิด {len(wrong)} รูป (แสดงสูงสุด 8 รูป) — ดูว่าผิดเพราะรูปไม่ชัด หรือโรคคล้ายกัน" if THAI_OK else "Misclassified", y=1.0)
    plt.tight_layout(); save_fig("05_misclassified"); plt.show()
else:
    ok("ไม่มีรูปที่ตอบผิดเลยในชุดสอบจริง")

# ---------- สรุปเกณฑ์ ----------
min_rec = rec[sup > 0].min() * 100 if (sup > 0).any() else 0
print("\n📋 สรุปความพร้อมใช้งาน")
if "จำลอง" in DATA_SOURCE:
    warn("ข้อมูลจำลอง: ตัวเลขนี้ไม่มีความหมายต่อการใช้งานจริง (ใช้ทดสอบระบบเท่านั้น)")
elif test_acc >= 90 and min_rec >= 80:
    ok(f"ผ่านเกณฑ์ดี: แม่นยำ {test_acc:.1f}% และทุกโรคจับได้ ≥ 80% (ต่ำสุด {min_rec:.1f}%) → ใช้เป็นเครื่องมือช่วยคัดกรองได้")
elif test_acc >= 80:
    warn(f"พอใช้: แม่นยำ {test_acc:.1f}% (โรคที่จับได้น้อยสุด {min_rec:.1f}%) → ควรเพิ่มรูปของโรคที่ Recall ต่ำ แล้วฝึกใหม่ก่อนใช้จริง")
else:
    warn(f"ยังไม่พร้อมใช้งาน: แม่นยำ {test_acc:.1f}% → เพิ่มรูปให้มากและหลากหลาย (มุม/แสง/ระยะ) แล้วฝึกใหม่")
if len(y_te) < 100:
    info(f"ชุดสอบจริงมีเพียง {len(y_te)} รูป ตัวเลขอาจคลาดเคลื่อนได้ ควรทดสอบกับรูปจากสวนจริงเพิ่ม")
````

## 🔍 ขั้นตอนที่ 11: ปัญญาประดิษฐ์ "มอง" ตรงไหน? (Grad-CAM)
นี่คือการ **ตรวจสอบว่าโมเดลตอบถูกด้วยเหตุผลที่ถูกต้อง** ไม่ใช่ตอบถูกเพราะฉากหลัง (เช่น ดิน มือ หรือแสงเงา)
ภาพ Grad-CAM จะซ้อนสีบนรูปใบ: 🔴 **สีแดง/เหลือง = จุดที่ปัญญาประดิษฐ์ใช้ตัดสินใจมาก** · 🔵 สีน้ำเงิน = ไม่สนใจ

**วิธีตรวจ:** ถ้าสีแดงอยู่ที่แผลบนใบ → เป็นสัญญาณที่ดี · ถ้าสีแดงอยู่ที่พื้นหลัง/ขอบรูป → โมเดลอาจเรียนทางลัดที่ผิด ควรเพิ่มรูปที่พื้นหลังหลากหลาย
> หมายเหตุ: เป็นเครื่องมือช่วยตีความ ไม่ใช่การวัดพื้นที่แผลที่แม่นยำ

````python
#@title 🔍 Grad-CAM: ดูว่าปัญญาประดิษฐ์มองตรงไหน { display-mode: "form" }
class GradCAM:
    # สร้างแผนที่ความร้อนจากชั้นสุดท้ายของส่วนที่ "มองภาพ" ของโมเดล
    def __init__(self, m, layer):
        self.m, self.act, self.grad = m, None, None
        self._h = layer.register_forward_hook(self._fwd)
    def _fwd(self, module, inp, out):
        self.act = out.detach()
        out.register_hook(lambda g: setattr(self, "grad", g.detach()))
    def close(self):
        self._h.remove()
    def __call__(self, x, target=None):
        self.m.eval(); self.m.zero_grad(set_to_none=True)
        x = x.clone().requires_grad_(True)
        logits = self.m(x)
        probs = F.softmax(logits.detach().float(), 1)[0].cpu().numpy()
        t = int(logits.argmax(1).item()) if target is None else int(target)
        logits[0, t].backward()
        w = self.grad.mean(dim=(2, 3), keepdim=True)
        cam = F.relu((w * self.act).sum(1))[0].float().cpu().numpy()
        cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)
        return cam, t, probs

def overlay_cam(pil_img, cam, alpha=0.45):
    h = Image.fromarray((cam * 255).astype(np.uint8)).resize(pil_img.size, Image.BILINEAR)
    heat = (matplotlib.colormaps["jet"](np.asarray(h) / 255.0)[..., :3] * 255).astype(np.uint8)
    base = np.asarray(pil_img.convert("RGB")).astype(np.float32)
    return np.clip((1 - alpha) * base + alpha * heat, 0, 255).astype(np.uint8)

def cam_for_image(pil_img):
    x = eval_tf(pil_img).unsqueeze(0).to(device)
    g = GradCAM(model, model.features[-1])
    try:
        cam, t, probs = g(x)
    finally:
        g.close()
    return cam, t, probs

# เลือกรูปที่ตอบถูกของแต่ละโรค 2 รูป (มั่นใจสุด) มาแสดง
PER = 2
fig, axes = plt.subplots(NUM_CLASSES, PER * 2, figsize=(3.0 * PER * 2, 3.1 * NUM_CLASSES), squeeze=False)
for ax in axes.flat: ax.axis("off")
for r in range(NUM_CLASSES):
    ok_idx = np.where((y_te == r) & (pred_te == r))[0]
    ok_idx = ok_idx[np.argsort(-probs_te[ok_idx, r])][:PER]
    for k, i in enumerate(ok_idx):
        pil = load_rgb(test_ds.paths[i], max_side=512)
        cam, t, _ = cam_for_image(pil)
        axes[r][2*k].imshow(pil.resize((IMAGE_SIZE, IMAGE_SIZE))); axes[r][2*k].set_title(th_name(class_names[r]) if k == 0 else "", loc="left", fontsize=10, color="darkgreen", weight="bold")
        axes[r][2*k+1].imshow(overlay_cam(pil.resize((IMAGE_SIZE, IMAGE_SIZE)), cam)); axes[r][2*k+1].set_title("จุดที่โมเดลมอง" if THAI_OK else "Grad-CAM", fontsize=9)
plt.suptitle("ซ้าย = รูปจริง · ขวา = จุดที่ปัญญาประดิษฐ์ใช้ตัดสิน (แดง = สำคัญมาก)" if THAI_OK else "Grad-CAM", y=1.0, fontsize=12)
plt.tight_layout(); save_fig("06_gradcam"); plt.show()
ok("ตรวจว่าสีแดงอยู่บนแผล/รอยโรค ไม่ใช่พื้นหลัง")
````

## 📱 ขั้นตอนที่ 13: ส่งออกไปใช้บนมือถือ/อุปกรณ์ในสวน (ONNX + TorchScript)
- **ONNX (`.onnx`)** — มาตรฐานสากล ใช้กับ Flutter, Android, iOS, เว็บ และ Raspberry Pi **ทำงานออฟไลน์ได้ 100%**
- **TorchScript (`.pt`)** — ใช้กับ PyTorch Mobile / C++

ไฟล์ที่ส่งออกจะให้ผลลัพธ์เป็น **เปอร์เซ็นต์ความน่าจะเป็นของแต่ละโรคโดยตรง** (ผ่าน softmax แล้ว) และระบบ **ทดสอบความถูกต้องของไฟล์ ONNX อัตโนมัติ** โดยเทียบกับโมเดลต้นฉบับ

````python
#@title 📱 ส่งออก ONNX / TorchScript + ทดสอบความถูกต้อง { display-mode: "form" }
class ExportWrapper(nn.Module):
    # ครอบโมเดลเพื่อให้ผลลัพธ์เป็นความน่าจะเป็น (softmax) พร้อมใช้ในแอป
    def __init__(self, m):
        super().__init__(); self.m = m
    def forward(self, x):
        return torch.softmax(self.m(x), dim=1)

export_model = ExportWrapper(copy.deepcopy(model).cpu().eval()).eval()
dummy = torch.randn(1, 3, IMAGE_SIZE, IMAGE_SIZE)
ONNX_PATH, TS_PATH = EXPORT_DIR / "durian_disease_model.onnx", EXPORT_DIR / "durian_disease_model.pt"
export_kw = dict(input_names=["input_leaf_image"], output_names=["probabilities"],
                 dynamic_axes={"input_leaf_image": {0: "batch"}, "probabilities": {0: "batch"}}, opset_version=17)

onnx_ok = False
for label, extra in [("ตัวส่งออกแบบเสถียร", {"dynamo": False}), ("ตัวส่งออกแบบใหม่", {"dynamo": True}), ("ตัวส่งออกรุ่นเก่า", {})]:
    try:
        torch.onnx.export(export_model, dummy, str(ONNX_PATH), **export_kw, **extra)
        onnx_ok = ONNX_PATH.exists() and ONNX_PATH.stat().st_size > 0
        if onnx_ok:
            ok(f"ส่งออก ONNX สำเร็จ ({label}): {ONNX_PATH}  ({ONNX_PATH.stat().st_size/1024**2:.1f} MB)"); break
    except Exception as e:
        info(f"{label}ใช้ไม่ได้ ({type(e).__name__}) ลองวิธีถัดไป ...")
if not onnx_ok:
    warn("ส่งออก ONNX ไม่สำเร็จ แต่ไฟล์ .pth ยังใช้งานได้ตามปกติ")

# ---------- ทดสอบว่า ONNX ตอบตรงกับโมเดลต้นฉบับ ----------
if onnx_ok:
    try:
        import onnxruntime as ort
        xb, _ = next(iter(test_loader)); xb = xb[:5].cpu()          # ใช้รูปจริง 5 รูป (ทดสอบ batch ≠ 1 ด้วย)
        sess = ort.InferenceSession(str(ONNX_PATH), providers=["CPUExecutionProvider"])
        out_onnx = sess.run(None, {sess.get_inputs()[0].name: xb.numpy()})[0]
        with torch.no_grad(): out_torch = export_model(xb).numpy()
        diff = float(np.abs(out_onnx - out_torch).max()); same = bool((out_onnx.argmax(1) == out_torch.argmax(1)).all())
        if diff < 1e-3 and same:
            ok(f"ทดสอบ ONNX ผ่าน: ผลต่างสูงสุดกับโมเดลต้นฉบับ {diff:.2e} และตอบตรงกันทุกรูป → พร้อมนำไปใส่แอป")
        else:
            warn(f"ONNX ให้ผลต่างจากต้นฉบับ (ต่างสูงสุด {diff:.2e}) ควรตรวจสอบก่อนใช้งานจริง")
    except Exception as e:
        warn(f"ทดสอบ ONNX อัตโนมัติไม่สำเร็จ ({type(e).__name__}: {e})")

# ---------- TorchScript ----------
try:
    ts = torch.jit.trace(export_model, dummy); ts.save(str(TS_PATH))
    ok(f"ส่งออก TorchScript สำเร็จ: {TS_PATH}  ({TS_PATH.stat().st_size/1024**2:.1f} MB)")
except Exception as e:
    info(f"ข้าม TorchScript ({type(e).__name__}) — ไม่กระทบไฟล์อื่น")
````

3. ดูผล: ชื่อโรค ความมั่นใจ จุดที่โมเดลมอง และคำแนะนำ

**เคล็ดลับถ่ายรูปให้ได้ผลดี:** ถ่ายใบเดียวให้เต็มเฟรม แสงธรรมชาติ ไม่เบลอ เห็นแผลชัด และพื้นหลังไม่รก
**รันช่องนี้ซ้ำได้ไม่จำกัด** ทุกครั้งที่อยากทดสอบรูปใหม่

````python
#@title 🧪 ตรวจโรคจากรูปที่อัปโหลด (รันซ้ำได้) { display-mode: "form" }
def diagnose(src, show=True):
    # src = ชื่อไฟล์รูป หรือ PIL.Image  → คืนค่าผลการตรวจเป็น dict
    pil = src.convert("RGB") if isinstance(src, Image.Image) else load_rgb(src, max_side=2048)
    cam, t, probs = cam_for_image(pil)
    cls = class_names[t]; conf = float(probs[t] * 100); I = info_of(cls)
    order = np.argsort(-probs); top2_gap = float((probs[order[0]] - probs[order[1]]) * 100) if NUM_CLASSES > 1 else 100.0

    if conf >= 85 and top2_gap >= 30: level, icon = "ความมั่นใจสูง", "🟢"
    elif conf >= 60:                  level, icon = "ความมั่นใจปานกลาง — ควรถ่ายซ้ำหรือให้ผู้เชี่ยวชาญยืนยัน", "🟡"
    else:                             level, icon = "ความมั่นใจต่ำ — ผลนี้ยังเชื่อถือไม่ได้", "🔴"

    if show:
        fig, ax = plt.subplots(1, 3, figsize=(14, 4.4), gridspec_kw={"width_ratios": [1, 1, 1.15]})
        small = pil.resize((IMAGE_SIZE, IMAGE_SIZE))
        ax[0].imshow(small); ax[0].set_title("รูปที่อัปโหลด" if THAI_OK else "input"); ax[0].axis("off")
        ax[1].imshow(overlay_cam(small, cam)); ax[1].set_title("จุดที่โมเดลมอง (แดง = สำคัญ)" if THAI_OK else "Grad-CAM"); ax[1].axis("off")
        names = [th_name(class_names[i]) for i in order][::-1]; vals = [probs[i] * 100 for i in order][::-1]
        bars = ax[2].barh(names, vals, color=["#bbbbbb"] * (len(vals) - 1) + ["#2e8b57"]); ax[2].set_xlim(0, 105)
        for b, v in zip(bars, vals): ax[2].text(v + 1, b.get_y() + b.get_height() / 2, f"{v:.1f}%", va="center", fontsize=9)
        ax[2].set_title("ความน่าจะเป็นของแต่ละโรค" if THAI_OK else "probabilities"); ax[2].spines[["top", "right"]].set_visible(False)
        plt.suptitle(f"ผลตรวจ: {th_name(cls)}  ({conf:.1f}%)" if THAI_OK else f"{cls} ({conf:.1f}%)", fontsize=14, color="darkgreen", weight="bold")
        plt.tight_layout(); plt.show()
        print("=" * 78)
        print(f"📋 ผลการตรวจ         : {I['th']}  ({cls})")
        print(f"{icon} ระดับความมั่นใจ    : {conf:.1f}%  → {level}")
        print(f"🦠 เชื้อสาเหตุ        : {I['pathogen']}")
        print(f"🌿 แนวทางจัดการ      : {I['guideline']}")
        if I["env"]:
            print(f"🌦️ สภาพที่ส่งเสริมโรค : ความชื้น {I['env']['trigger_rh']} | อุณหภูมิ {I['env']['trigger_temp']} | {I['env']['risk_season']}")
        print("⚠️ โมเดลรู้จักเฉพาะกลุ่มต่อไปนี้: " + ", ".join(th_name(c) for c in class_names))
        print("   หากรูปไม่ใช่ใบทุเรียน หรือเป็นโรค/ปัญหาอื่น (เช่น ขาดธาตุอาหาร แมลง) ผลอาจผิดพลาด ควรปรึกษาผู้เชี่ยวชาญ")
        print("=" * 78)
    return {"class": cls, "thai": I["th"], "confidence": conf, "probabilities": {class_names[i]: float(probs[i] * 100) for i in range(NUM_CLASSES)}}

if IN_COLAB:
    from google.colab import files
    print("👇 กดปุ่ม 'Choose Files' เพื่อเลือกรูปใบทุเรียน (เลือกได้หลายรูป)")
    uploaded = files.upload()
    if not uploaded:
        info("ยังไม่ได้เลือกรูป — กดรันช่องนี้ใหม่เมื่อพร้อม")
    for fname in uploaded:
        print(f"\n🖼️ ไฟล์: {fname}")
        try:
            diagnose(fname)
        except Exception as e:
            warn(f"เปิดรูป '{fname}' ไม่ได้ ({type(e).__name__}) — ตรวจว่าเป็นไฟล์รูป .jpg/.png")
else:
    info("ไม่ได้อยู่บน Colab: เรียกใช้ด้วย diagnose('path/to/leaf.jpg') ได้เลย")
    _demo = diagnose(test_ds.paths[0])      # ตัวอย่างการตรวจด้วยรูปจากชุดสอบจริง
````
